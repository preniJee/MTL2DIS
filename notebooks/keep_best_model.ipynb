```python
import os
import json
from shutil import copyfile, rmtree, copytree
import sys
sys.path.append('../')
from utils import get_result_path
```

```python


source_dir ="../results/roberta-base/brexit/Hate/seed_0/budget_2352/mtl_Ann1,Ann2,Ann3,Ann4,Ann5,Ann6_32/hyperparam_tuning"
destination_dir = "../results/roberta-base/brexit/Hate/seed_0/budget_2352/mtl_Ann1,Ann2,Ann3,Ann4,Ann5,Ann6_32"


best_model_path = None
best_avg_val_f1 = float('-inf')

# Traverse through the subdirectories
for current_directory, subdirectories, files in os.walk(source_dir):
        # Check if there is a "model" file in the current directory
        if 'model' in subdirectories:
            # Load the "test_result.json" file
            test_result_path = os.path.join(current_directory, 'val_result.json')
            print(test_result_path)
            if os.path.exists(test_result_path):
                with open(test_result_path, 'r') as f:
                    test_result_data = json.load(f)

                # Extract the relevant metric (e.g., avg_val_f1)
                avg_val_f1 = test_result_data.get('avg_val_f1', 0.0)

                # Update the best model if needed
                if avg_val_f1 > best_avg_val_f1:
                    best_avg_val_f1 = avg_val_f1
                    best_model_path = current_directory

# # Copy the best model and "test_result.json" to the destination directory
if best_model_path:
    copytree(os.path.join(best_model_path, 'model'), os.path.join(destination_dir, 'best_model'))
    copyfile(os.path.join(best_model_path, 'val_result.json'), os.path.join(destination_dir, 'val_result.json'))

# Delete subdirectories and their contents
rmtree(source_dir)
```
