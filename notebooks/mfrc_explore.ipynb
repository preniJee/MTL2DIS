```python
import pandas as pd
df = pd.read_csv("../data/mfrc/mini_mfrc.csv", index_col=0)
df.head()
```

```python
df.columns
```

```python
df_filtered = df[['text_id', 'annotator_id', 'text', 'Non-Moral']]
df_filtered['Moral'] = df_filtered['Non-Moral'].apply(lambda x: 1 if x == 0 else 0)
df_filtered.head()
df_filtered.groupby('annotator_id')['Moral'].value_counts().reset_index(name='count')
```

```python
df_filtered.groupby('annotator_id').count().reset_index().sort_values(by='text_id', ascending=False)
# df_filtered['annotator_id'].nunique()
```

```python
exclude = ['aaronvalencia13', 'AAronvalencia13', 'victoria']
df_filtered = df_filtered[~df_filtered['annotator_id'].isin(exclude)]
df_filtered['annotator_id'].nunique()
df_filtered = df_filtered[['text_id', 'annotator_id', 'text', 'Moral']]
df_filtered['ann_id'] = pd.factorize(df_filtered['annotator_id'])[0]
```

### split train test dev and save for each annotator

```python
from sklearn.model_selection import train_test_split
seed = 0

text_ids = df_filtered['text_id'].unique().tolist()
train_ids, test_ids = train_test_split(text_ids, test_size=0.2, random_state=seed)
val_ids, test_ids = train_test_split(test_ids, test_size=0.5, random_state=seed)
len(train_ids), len(test_ids), len(val_ids)
```

```python
train_df = df_filtered[df_filtered['text_id'].isin(train_ids)]
val_df = df_filtered[df_filtered['text_id'].isin(val_ids)]
test_df = df_filtered[df_filtered['text_id'].isin(test_ids)]

train_df.groupby('annotator_id')['Moral'].value_counts().reset_index(name='count')
```

```python
import os
label = 'Moral'
path = f'../data/mfrc/{label}/annotators'
os.makedirs(path, exist_ok=True)
for ann in df_filtered['ann_id'].unique():
    ann_path = f'{path}/{ann}'
    os.makedirs(ann_path, exist_ok=True)
    df_train = train_df[train_df['ann_id'] == ann]
    df_val = val_df[val_df['ann_id'] == ann]
    df_test = test_df[test_df['ann_id'] == ann]
    print(ann , len(df_train), len(df_val), len(df_test))
    # df_train.to_csv(f'{path}/{ann}/train.csv')
    # df_val.to_csv(f'{path}/{ann}/val.csv')
    # df_test.to_csv(f'{path}/{ann}/test.csv')
    
```
